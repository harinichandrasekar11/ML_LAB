```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully")
```

```python
books = pd.read_csv("books.csv")

print("Books Dataset:")
print(books.head())

print("\nColumns:")
print(books.columns.tolist())
```

```python
ratings = pd.read_csv(
    "ratings.csv",
    sep="|",
    header=None,
    names=["User", "Title", "Rating"]
)

print("Ratings Dataset:")
print(ratings.head())

print("\nColumns:")
print(ratings.columns.tolist())
```

```python
print("Number of books:", len(books))
print("Number of ratings:", len(ratings))

print("\nBooks:")
display(books.head())

print("\nRatings:")
display(ratings.head())
```

```python
book_matrix = ratings.pivot_table(
    index="Title",
    columns="User",
    values="Rating"
)

book_matrix = book_matrix.fillna(0)

print("User-Book Matrix:")
display(book_matrix.head())
```

```python
similarity = cosine_similarity(book_matrix)

print("Cosine similarity calculated successfully!")
print("Matrix shape:", similarity.shape)
```

```python
def recommend_books(book_name, n=5):
    
    if book_name not in book_matrix.index:
        print("Book not found in ratings dataset.")
        return
    
    book_index = book_matrix.index.get_loc(book_name)
    
    scores = list(enumerate(similarity[book_index]))
    
    scores = sorted(
        scores,
        key=lambda x: x[1],
        reverse=True
    )
    
    print("\nRecommended Books:")
    
    count = 0
    
    for index, score in scores[1:]:
        print(book_matrix.index[index])
        
        count += 1
        
        if count == n:
            break
```

```python
print("Sample book names:\n")

for book in book_matrix.index[:20]:
    print(book)
```

```python
recommend_books("Kolya (1996)", 5)
```
